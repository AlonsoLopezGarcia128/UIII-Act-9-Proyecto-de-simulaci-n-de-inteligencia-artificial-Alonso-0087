```python
# --- SIMULADOR DE IA: ASISTENTE CIENTÍFICO 'GENIUS-LAB 2026' ---

# Inicialización de la bitácora vacía para registrar hallazgos (Source 47, 48)
descubrimientos = [] # Crea una lista vacía para almacenar cadenas de texto
ejecutando = True # Variable de control booleana para mantener el ciclo activo (Source 39)

# Bienvenida interactiva al usuario (Source 59)
print("INICIALIZANDO PROTOCOLOS DE INTELIGENCIA ARTIFICIAL...")
nombre = input("Científico, ingrese su identificación: ") # Recibe el nombre del usuario
print(f"Sesión iniciada. Bienvenido al laboratorio, Dr/Dra. {nombre}.")
print() # Genera un espacio visual para mejorar la legibilidad

# Inicio del ciclo principal tipo menú (Source 62, 559)
while ejecutando == True: # El bucle se repite mientras la condición sea verdadera
    print("--- MENÚ DE COMANDOS IA ---")
    print("1. Registrar nuevo descubrimiento") # Uso de listas
    print("2. Análisis de seguridad de muestra (pH)") # Operadores lógicos
    print("3. Calculadora de densidad de reactivos") # Operadores aritméticos
    print("4. Revisar bitácora de hallazgos") # Acceso a listas
    print("5. Desactivar IA") # Salida del sistema

    opcion = input("Seleccione una operación (1-5): ") # Solicita la opción al usuario
    print() # Espaciado solicitado tras la respuesta

    # Lógica de decisión múltiple (Source 58, 486, 1328)
    if opcion == "1": # Compara si la opción elegida es igual a "1" (Relacional)
        hallazgo = input("Describa el nuevo descubrimiento: ") # Recibe el dato
        descubrimientos.append(hallazgo) # Método append para añadir a la lista (Source 48)
        print("SISTEMA: Hallazgo registrado exitosamente en la bitácora.")
        print() # Espaciado

    elif opcion == "2": # Evalúa otra condición si la primera fue falsa (Source 58)
        ph = float(input("Ingrese el nivel de pH de la sustancia (0-14): ")) # Convierte a decimal
        # Uso de operadores lógicos 'and' para determinar rangos (Source 59, 437)
        if ph >= 0 and ph < 7: # Verifica si es un ácido
            print("IA DIAGNÓSTICO: La sustancia es un Ácido. ¡Manejar con guantes!")
        elif ph > 7 and ph <= 14: # Verifica si es una base
            print("IA DIAGNÓSTICO: La sustancia es Alcalina (Base).")
        elif ph == 7: # Operador de igualdad exacta (Source 482)
            print("IA DIAGNÓSTICO: La sustancia es Neutra (Agua Pura).")
        else: # Caso para valores fuera de la escala (Source 58)
            print("IA ERROR: El valor ingresado no pertenece a la escala de pH.")
        print() # Espaciado

    elif opcion == "3":
        masa = float(input("Ingrese la masa del reactivo (gramos): ")) # Entrada de masa
        volumen = float(input("Ingrese el volumen ocupado (ml): ")) # Entrada de volumen
        # Operador aritmético de división '/' y relacional '!=' (Source 41, 482)
        if volumen != 0: # Verifica que el volumen no sea cero para evitar errores
            densidad = masa / volumen # Realiza el cálculo matemático (Source 157)
            print(f"IA CÁLCULO: La densidad resultante es {densidad:.2f} g/ml.")
        else:
            print("IA ERROR: No se puede calcular densidad con volumen cero.")
        print() # Espaciado

    elif opcion == "4":
        print("--- BITÁCORA DE DESCUBRIMIENTOS ---")
        # Uso de operadores relacionales para verificar si la lista tiene datos (Source 51)
        if len(descubrimientos) > 0: # La función len devuelve el tamaño de la lista (Source 49)
            print(f"Total de registros: {len(descubrimientos)}")
            # Al no poder usar 'for', mostramos el último registro como ejemplo de acceso
            print(f"Último hallazgo registrado: {descubrimientos[-1]}") # Acceso por índice (Source 48)
        else:
            print("SISTEMA: La bitácora está actualmente vacía.")
        print() # Espaciado

    elif opcion == "5":
        print(f"Desconectando GENIUS-LAB 2026. ¡Hasta pronto, {nombre}!")
        ejecutando = False # Cambia el valor para romper el ciclo while (Source 64)

    else: # Manejo de opciones no válidas
        print("IA AVISO: Comando no reconocido. Por favor, intente de nuevo.")
        print() # Espaciado
```
